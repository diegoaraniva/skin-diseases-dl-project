# Proyecto Final DL — Etapa 1 (Línea Base)

## Propósito de la etapa
La Etapa 1 establece la línea base metodológica y cuantitativa del proyecto de clasificación multiclase de enfermedades dermatológicas. Su objetivo es construir un punto de referencia sólido para comparar, de manera trazable, las mejoras introducidas en etapas posteriores.

## Alcance metodológico
La etapa cubre de forma integral los componentes fundamentales del flujo de modelado supervisado:

1. Selección y documentación del dataset.
2. Análisis exploratorio de datos (EDA) orientado a imágenes.
3. Preparación y partición estratificada de datos para entrenamiento, validación y prueba.
4. Implementación de un modelo base neuronal (MLP multiclase).
5. Evaluación formal con métricas de clasificación y análisis complementario.

## Criterios de evaluación
La evaluación se fundamenta en métricas de clasificación multiclase: **Accuracy**, **Precision macro**, **Recall macro** y **Macro F1**. La meta de referencia de 0.60 se usa como umbral orientativo de desempeño inicial, no como criterio único de validez metodológica.

## Dataset seleccionado
- Fuente: Kaggle
- Nombre: Skin Diseases Cancer Comprehensive Dataset
- Croissant JSON-LD:
  https://www.kaggle.com/datasets/anieetorudofia/skin-diseases-cancer-comprehensive-dataset/croissant/download

El conjunto de datos está compuesto principalmente por imágenes organizadas por clase. Por esta razón, el análisis se enfoca en estructura de carpetas, etiquetas y metadatos de imagen, en lugar de variables tabulares tradicionales.

```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from PIL import Image

tf.get_logger().setLevel('ERROR')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
)
from sklearn.utils.class_weight import compute_class_weight

from tensorflow.keras import mixed_precision
```

```python
SEED = 42
np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)


mixed_precision.set_global_policy('mixed_float16')
print('Mixed precision policy:', mixed_precision.global_policy())
print('GPUs detectadas:', tf.config.list_physical_devices('GPU'))


valid_ext = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff', '.webp'}

kaggle_input_root = Path('/kaggle/input')
if not kaggle_input_root.exists():
    raise FileNotFoundError('No se encontró /kaggle/input. Este bloque está preparado para ejecutarse en Kaggle.')


def count_images_in_dir(root_dir: Path) -> int:
    return sum(1 for p in root_dir.rglob('*') if p.is_file() and p.suffix.lower() in valid_ext)


candidates = []
for child in kaggle_input_root.iterdir():
    if child.is_dir():
        n_imgs = count_images_in_dir(child)
        if n_imgs > 0:
            candidates.append((n_imgs, child))

if not candidates:
    raise ValueError('No se detectaron datasets de imágenes dentro de /kaggle/input.')


dataset_path = max(candidates, key=lambda x: x[0])[1]
print('Dataset detectado:', dataset_path)
print('Total imágenes detectadas en ese dataset:', max(candidates, key=lambda x: x[0])[0])
```

```python
image_files = [
    p for p in dataset_path.rglob('*')
    if p.is_file() and p.suffix.lower() in valid_ext
]


images_df = pd.DataFrame({
    'file_path': [str(p) for p in image_files],
    'label': [p.parent.name for p in image_files],
    'ext': [p.suffix.lower() for p in image_files],
    'file_size_bytes': [p.stat().st_size for p in image_files],
})


print('Total imágenes encontradas:', len(images_df))
print('Clases detectadas:', images_df['label'].nunique())
display(images_df.head())
```

```python
def read_image_metadata(path_str: str):
    try:
        with Image.open(path_str) as img:
            width, height = img.size
            mode = img.mode
        channels_map = {"L": 1, "RGB": 3, "RGBA": 4}
        channels = channels_map.get(mode, np.nan)
        return width, height, mode, channels, None
    except Exception as err:
        return np.nan, np.nan, None, np.nan, str(err)

meta = images_df["file_path"].apply(read_image_metadata)
images_df[["width", "height", "mode", "channels", "read_error"]] = pd.DataFrame(meta.tolist(), index=images_df.index)
images_df["aspect_ratio"] = images_df["width"] / images_df["height"]
images_df["file_size_kb"] = images_df["file_size_bytes"] / 1024

images_df.head()
```

```python
df = images_df.copy()

print('Total de imágenes:', df.shape[0])
print('Total de clases:', df['label'].nunique())
print('Variables disponibles:', list(df.columns))
print('Duplicados por ruta:', df.duplicated(subset=['file_path']).sum())
print('Errores de lectura:', df['read_error'].notna().sum())

print('\nTop 10 clases por frecuencia:')
display(df['label'].value_counts().head(10).to_frame('count'))

print('\nResumen de faltantes por variable:')
display(df.isna().sum().to_frame('missing').sort_values('missing', ascending=False))
```

```python


sns.set_theme(style='whitegrid')

class_dist = df['label'].value_counts()

plt.figure(figsize=(13, 5))
sns.barplot(x=class_dist.head(20).index, y=class_dist.head(20).values)
plt.title('Top 20 clases por número de imágenes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

for col in ['width', 'height', 'file_size_kb', 'aspect_ratio']:
    plt.figure(figsize=(7, 4))
    sns.histplot(df[col].dropna(), bins=40, kde=True)
    plt.title(f'Distribución de {col}')
    plt.tight_layout()
    plt.show()
```

```python

IMG_SIZE_F1 = (192, 192)
BATCH_SIZE_F1 = 32
EPOCHS_F1 = 35
LR_F1 = 6e-4
WEIGHT_DECAY_F1 = 2e-4






def build_phase1_base_mlp(num_classes: int):
    inputs = tf.keras.layers.Input(shape=(IMG_SIZE_F1[0], IMG_SIZE_F1[1], 3))

    
    x = tf.keras.layers.Rescaling(1.0 / 255.0)(inputs)

    
    x = tf.keras.layers.RandomFlip('horizontal')(x)
    x = tf.keras.layers.RandomRotation(0.10)(x)
    x = tf.keras.layers.RandomZoom(0.12)(x)
    x = tf.keras.layers.RandomContrast(0.12)(x)

    
    x = tf.keras.layers.AveragePooling2D(pool_size=6)(x)
    x = tf.keras.layers.Flatten()(x)

    
    x = tf.keras.layers.Dense(768, activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(0.45)(x)

    x = tf.keras.layers.Dense(384, activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(0.35)(x)

    x = tf.keras.layers.Dense(192, activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(0.25)(x)

    
    outputs = tf.keras.layers.Dense(num_classes, activation='softmax', dtype='float32')(x)

    model = tf.keras.Model(inputs, outputs, name='phase1_base_mlp')
    return model

print('Modelo Fase 1 configurado: MLP base.')
```

```python

AUTOTUNE = tf.data.AUTOTUNE





phase1_df = df[['file_path', 'label']].dropna().copy()
class_counts = phase1_df['label'].value_counts()
phase1_df = phase1_df[phase1_df['label'].isin(class_counts[class_counts >= 5].index)].reset_index(drop=True)


label_encoder_f1 = LabelEncoder()
phase1_df['target'] = label_encoder_f1.fit_transform(phase1_df['label'])
num_classes_f1 = len(label_encoder_f1.classes_)


train_df, temp_df = train_test_split(
    phase1_df,
    test_size=0.30,
    random_state=SEED,
    stratify=phase1_df['target']
)
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_df['target']
)

print('Split Fase 1 -> train/val/test:', len(train_df), len(val_df), len(test_df))
print('Clases usadas en Fase 1:', num_classes_f1)


classes = np.unique(train_df['target'].values)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=train_df['target'].values)
weights = np.sqrt(weights)
class_weight_f1 = {int(c): float(w) for c, w in zip(classes, weights)}





def decode_and_resize(path, label):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE_F1, antialias=True)
    img = tf.cast(img, tf.float32)
    return img, label

def make_ds(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(buffer_size=min(len(paths), 12000), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(decode_and_resize, num_parallel_calls=AUTOTUNE)
    ds = ds.batch(BATCH_SIZE_F1)
    ds = ds.prefetch(AUTOTUNE)
    return ds

train_ds = make_ds(train_df['file_path'].values, train_df['target'].values, training=True)
val_ds = make_ds(val_df['file_path'].values, val_df['target'].values, training=False)
test_ds = make_ds(test_df['file_path'].values, test_df['target'].values, training=False)




phase1_model = build_phase1_base_mlp(num_classes=num_classes_f1)

callbacks_f1 = [
    
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True),
    
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=5e-7),
]


optimizer_f1 = tf.keras.optimizers.AdamW(
    learning_rate=LR_F1,
    weight_decay=WEIGHT_DECAY_F1,
    clipnorm=1.0
)

phase1_model.compile(
    optimizer=optimizer_f1,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_f1 = phase1_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_F1,
    class_weight=class_weight_f1,
    callbacks=callbacks_f1,
    verbose=1,
)

y_test_f1 = test_df['target'].values
y_prob_f1 = phase1_model.predict(test_ds, verbose=0)
y_pred_f1 = np.argmax(y_prob_f1, axis=1)

acc_eval = accuracy_score(y_test_f1, y_pred_f1)
precision_macro, recall_macro, f1_macro_eval, _ = precision_recall_fscore_support(
    y_test_f1,
    y_pred_f1,
    average='macro',
    zero_division=0,
)

metrics_eval_df = pd.DataFrame({
    'metric': ['accuracy', 'precision_macro', 'recall_macro', 'macro_f1', 'target_goal'],
    'value': [acc_eval, precision_macro, recall_macro, f1_macro_eval, 0.60],
})

print(f'Fase 1 - accuracy: {acc_eval:.4f}')
print(f'Fase 1 - precision_macro: {precision_macro:.4f}')
print(f'Fase 1 - recall_macro: {recall_macro:.4f}')
print(f'Fase 1 - macro F1: {f1_macro_eval:.4f}')
print(f'Objetivo >= 0.60 cumplido: {acc_eval >= 0.60 and f1_macro_eval >= 0.60}')

if acc_eval < 0.60 or f1_macro_eval < 0.60:
    print('Resultado por debajo de la meta de referencia (0.60).')

display(metrics_eval_df)

plt.figure(figsize=(8, 4))
sns.barplot(data=metrics_eval_df, x='metric', y='value')
plt.axhline(0.60, color='green', linestyle='--', label='Objetivo 0.60')
plt.ylim(0, 1)
plt.title('Fase 1 (Kaggle) - Métricas de evaluación')
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history_f1.history['loss'], label='train_loss')
plt.plot(history_f1.history['val_loss'], label='val_loss')
plt.title('Curvas de pérdida')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_f1.history['accuracy'], label='train_accuracy')
plt.plot(history_f1.history['val_accuracy'], label='val_accuracy')
plt.title('Curvas de accuracy')
plt.legend()
plt.tight_layout()
plt.show()




mis_idx = np.where(y_pred_f1 != y_test_f1)[0]
print('Total de errores en test:', len(mis_idx))

if len(mis_idx) > 0:
    mis_df = test_df.iloc[mis_idx].copy().reset_index(drop=True)
    mis_df['true_label'] = label_encoder_f1.inverse_transform(y_test_f1[mis_idx])
    mis_df['pred_label'] = label_encoder_f1.inverse_transform(y_pred_f1[mis_idx])
    mis_df['pred_confidence'] = y_prob_f1[mis_idx].max(axis=1)

    display(mis_df[['file_path', 'true_label', 'pred_label', 'pred_confidence']].head(10))

    n_show = min(6, len(mis_df))
    sample_mis = mis_df.sample(n=n_show, random_state=SEED).reset_index(drop=True)
    plt.figure(figsize=(14, 7))
    for i in range(n_show):
        img = tf.keras.utils.load_img(sample_mis.loc[i, 'file_path'], target_size=IMG_SIZE_F1)
        plt.subplot(2, 3, i + 1)
        plt.imshow(img)
        plt.title(
            f"Real: {sample_mis.loc[i, 'true_label']}\nPred: {sample_mis.loc[i, 'pred_label']}\nConf: {sample_mis.loc[i, 'pred_confidence']:.2f}"
        )
        plt.axis('off')
    plt.tight_layout()
    plt.show()

print('\nReporte de clasificación (macro):')
print(classification_report(y_test_f1, y_pred_f1, target_names=label_encoder_f1.classes_, zero_division=0))

cm = confusion_matrix(y_test_f1, y_pred_f1)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, cmap='Blues', cbar=False)
plt.title('Matriz de confusión (Fase 1)')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.show()
```

## Métricas de evaluación definidas

Para esta Fase 1, formulada como clasificación multiclase, se reportan las métricas principales de desempeño:

- Recall

Aclaración metodológica: RMSE y MAE corresponden a problemas de regresión, por lo que no se consideran métricas pertinentes para este caso de clasificación de imágenes.
