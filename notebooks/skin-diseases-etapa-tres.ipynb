# Proyecto Final DL — Etapa 1 (Línea Base)

## Propósito de la etapa
La Etapa 1 establece la línea base metodológica y cuantitativa del proyecto de clasificación multiclase de enfermedades dermatológicas. Su objetivo es construir un punto de referencia sólido para comparar, de manera trazable, las mejoras introducidas en etapas posteriores.

## Alcance metodológico
La etapa cubre de forma integral los componentes fundamentales del flujo de modelado supervisado:

1. Selección y documentación del dataset.
2. Análisis exploratorio de datos (EDA) orientado a imágenes.
3. Preparación y partición estratificada de datos para entrenamiento, validación y prueba.
4. Implementación de un modelo base neuronal (MLP multiclase).
5. Evaluación formal con métricas de clasificación y análisis complementario.

## Criterios de evaluación
La evaluación se fundamenta en métricas de clasificación multiclase: **Accuracy**, **Precision macro**, **Recall macro** y **Macro F1**. La meta de referencia de 0.60 se usa como umbral orientativo de desempeño inicial, no como criterio único de validez metodológica.

## Dataset seleccionado
- Fuente: Kaggle
- Nombre: Skin Diseases Cancer Comprehensive Dataset
- Croissant JSON-LD:
  https://www.kaggle.com/datasets/anieetorudofia/skin-diseases-cancer-comprehensive-dataset/croissant/download

El conjunto de datos está compuesto principalmente por imágenes organizadas por clase. Por esta razón, el análisis se enfoca en estructura de carpetas, etiquetas y metadatos de imagen, en lugar de variables tabulares tradicionales.

```python
import os
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from PIL import Image

tf.get_logger().setLevel('ERROR')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
)
from sklearn.utils.class_weight import compute_class_weight

from tensorflow.keras import mixed_precision
```

```python
SEED = 42
np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)


mixed_precision.set_global_policy('mixed_float16')
print('Mixed precision policy:', mixed_precision.global_policy())
print('GPUs detectadas:', tf.config.list_physical_devices('GPU'))


valid_ext = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff', '.webp'}

kaggle_input_root = Path('/kaggle/input')
if not kaggle_input_root.exists():
    raise FileNotFoundError('No se encontró /kaggle/input. Este bloque está preparado para ejecutarse en Kaggle.')


def count_images_in_dir(root_dir: Path) -> int:
    return sum(1 for p in root_dir.rglob('*') if p.is_file() and p.suffix.lower() in valid_ext)


candidates = []
for child in kaggle_input_root.iterdir():
    if child.is_dir():
        n_imgs = count_images_in_dir(child)
        if n_imgs > 0:
            candidates.append((n_imgs, child))

if not candidates:
    raise ValueError('No se detectaron datasets de imágenes dentro de /kaggle/input.')


dataset_path = max(candidates, key=lambda x: x[0])[1]
print('Dataset detectado:', dataset_path)
print('Total imágenes detectadas en ese dataset:', max(candidates, key=lambda x: x[0])[0])
```

```python
image_files = [
    p for p in dataset_path.rglob('*')
    if p.is_file() and p.suffix.lower() in valid_ext
]


images_df = pd.DataFrame({
    'file_path': [str(p) for p in image_files],
    'label': [p.parent.name for p in image_files],
    'ext': [p.suffix.lower() for p in image_files],
    'file_size_bytes': [p.stat().st_size for p in image_files],
})


print('Total imágenes encontradas:', len(images_df))
print('Clases detectadas:', images_df['label'].nunique())
display(images_df.head())
```

```python
def read_image_metadata(path_str: str):
    try:
        with Image.open(path_str) as img:
            width, height = img.size
            mode = img.mode
        channels_map = {"L": 1, "RGB": 3, "RGBA": 4}
        channels = channels_map.get(mode, np.nan)
        return width, height, mode, channels, None
    except Exception as err:
        return np.nan, np.nan, None, np.nan, str(err)

meta = images_df["file_path"].apply(read_image_metadata)
images_df[["width", "height", "mode", "channels", "read_error"]] = pd.DataFrame(meta.tolist(), index=images_df.index)
images_df["aspect_ratio"] = images_df["width"] / images_df["height"]
images_df["file_size_kb"] = images_df["file_size_bytes"] / 1024

images_df.head()
```

```python
df = images_df.copy()

print('Total de imágenes:', df.shape[0])
print('Total de clases:', df['label'].nunique())
print('Variables disponibles:', list(df.columns))
print('Duplicados por ruta:', df.duplicated(subset=['file_path']).sum())
print('Errores de lectura:', df['read_error'].notna().sum())

print('\nTop 10 clases por frecuencia:')
display(df['label'].value_counts().head(10).to_frame('count'))

print('\nResumen de faltantes por variable:')
display(df.isna().sum().to_frame('missing').sort_values('missing', ascending=False))
```

```python


sns.set_theme(style='whitegrid')

class_dist = df['label'].value_counts()

plt.figure(figsize=(13, 5))
sns.barplot(x=class_dist.head(20).index, y=class_dist.head(20).values)
plt.title('Top 20 clases por número de imágenes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

for col in ['width', 'height', 'file_size_kb', 'aspect_ratio']:
    plt.figure(figsize=(7, 4))
    sns.histplot(df[col].dropna(), bins=40, kde=True)
    plt.title(f'Distribución de {col}')
    plt.tight_layout()
    plt.show()
```

```python

IMG_SIZE_F1 = (192, 192)
BATCH_SIZE_F1 = 32
EPOCHS_F1 = 35
LR_F1 = 6e-4
WEIGHT_DECAY_F1 = 2e-4






def build_phase1_base_mlp(num_classes: int):
    inputs = tf.keras.layers.Input(shape=(IMG_SIZE_F1[0], IMG_SIZE_F1[1], 3))

    
    x = tf.keras.layers.Rescaling(1.0 / 255.0)(inputs)

    
    x = tf.keras.layers.RandomFlip('horizontal')(x)
    x = tf.keras.layers.RandomRotation(0.10)(x)
    x = tf.keras.layers.RandomZoom(0.12)(x)
    x = tf.keras.layers.RandomContrast(0.12)(x)

    
    x = tf.keras.layers.AveragePooling2D(pool_size=6)(x)
    x = tf.keras.layers.Flatten()(x)

    
    x = tf.keras.layers.Dense(768, activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(0.45)(x)

    x = tf.keras.layers.Dense(384, activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(0.35)(x)

    x = tf.keras.layers.Dense(192, activation='relu')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Dropout(0.25)(x)

    
    outputs = tf.keras.layers.Dense(num_classes, activation='softmax', dtype='float32')(x)

    model = tf.keras.Model(inputs, outputs, name='phase1_base_mlp')
    return model

print('Modelo Fase 1 configurado: MLP base.')
```

```python

AUTOTUNE = tf.data.AUTOTUNE





phase1_df = df[['file_path', 'label']].dropna().copy()
class_counts = phase1_df['label'].value_counts()
phase1_df = phase1_df[phase1_df['label'].isin(class_counts[class_counts >= 5].index)].reset_index(drop=True)


label_encoder_f1 = LabelEncoder()
phase1_df['target'] = label_encoder_f1.fit_transform(phase1_df['label'])
num_classes_f1 = len(label_encoder_f1.classes_)


train_df, temp_df = train_test_split(
    phase1_df,
    test_size=0.30,
    random_state=SEED,
    stratify=phase1_df['target']
)
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_df['target']
)

print('Split Fase 1 -> train/val/test:', len(train_df), len(val_df), len(test_df))
print('Clases usadas en Fase 1:', num_classes_f1)


classes = np.unique(train_df['target'].values)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=train_df['target'].values)
weights = np.sqrt(weights)
class_weight_f1 = {int(c): float(w) for c, w in zip(classes, weights)}





def decode_and_resize(path, label):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE_F1, antialias=True)
    img = tf.cast(img, tf.float32)
    return img, label

def make_ds(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(buffer_size=min(len(paths), 12000), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(decode_and_resize, num_parallel_calls=AUTOTUNE)
    ds = ds.batch(BATCH_SIZE_F1)
    ds = ds.prefetch(AUTOTUNE)
    return ds

train_ds = make_ds(train_df['file_path'].values, train_df['target'].values, training=True)
val_ds = make_ds(val_df['file_path'].values, val_df['target'].values, training=False)
test_ds = make_ds(test_df['file_path'].values, test_df['target'].values, training=False)




phase1_model = build_phase1_base_mlp(num_classes=num_classes_f1)

callbacks_f1 = [
    
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True),
    
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=5e-7),
]


optimizer_f1 = tf.keras.optimizers.AdamW(
    learning_rate=LR_F1,
    weight_decay=WEIGHT_DECAY_F1,
    clipnorm=1.0
)

phase1_model.compile(
    optimizer=optimizer_f1,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_f1 = phase1_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS_F1,
    class_weight=class_weight_f1,
    callbacks=callbacks_f1,
    verbose=1,
)

y_test_f1 = test_df['target'].values
y_prob_f1 = phase1_model.predict(test_ds, verbose=0)
y_pred_f1 = np.argmax(y_prob_f1, axis=1)

acc_eval = accuracy_score(y_test_f1, y_pred_f1)
precision_macro, recall_macro, f1_macro_eval, _ = precision_recall_fscore_support(
    y_test_f1,
    y_pred_f1,
    average='macro',
    zero_division=0,
)

metrics_eval_df = pd.DataFrame({
    'metric': ['accuracy', 'precision_macro', 'recall_macro', 'macro_f1', 'target_goal'],
    'value': [acc_eval, precision_macro, recall_macro, f1_macro_eval, 0.60],
})

print(f'Fase 1 - accuracy: {acc_eval:.4f}')
print(f'Fase 1 - precision_macro: {precision_macro:.4f}')
print(f'Fase 1 - recall_macro: {recall_macro:.4f}')
print(f'Fase 1 - macro F1: {f1_macro_eval:.4f}')
print(f'Objetivo >= 0.60 cumplido: {acc_eval >= 0.60 and f1_macro_eval >= 0.60}')

if acc_eval < 0.60 or f1_macro_eval < 0.60:
    print('Resultado por debajo de la meta de referencia (0.60).')

display(metrics_eval_df)

plt.figure(figsize=(8, 4))
sns.barplot(data=metrics_eval_df, x='metric', y='value')
plt.axhline(0.60, color='green', linestyle='--', label='Objetivo 0.60')
plt.ylim(0, 1)
plt.title('Fase 1 (Kaggle) - Métricas de evaluación')
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history_f1.history['loss'], label='train_loss')
plt.plot(history_f1.history['val_loss'], label='val_loss')
plt.title('Curvas de pérdida')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_f1.history['accuracy'], label='train_accuracy')
plt.plot(history_f1.history['val_accuracy'], label='val_accuracy')
plt.title('Curvas de accuracy')
plt.legend()
plt.tight_layout()
plt.show()




mis_idx = np.where(y_pred_f1 != y_test_f1)[0]
print('Total de errores en test:', len(mis_idx))

if len(mis_idx) > 0:
    mis_df = test_df.iloc[mis_idx].copy().reset_index(drop=True)
    mis_df['true_label'] = label_encoder_f1.inverse_transform(y_test_f1[mis_idx])
    mis_df['pred_label'] = label_encoder_f1.inverse_transform(y_pred_f1[mis_idx])
    mis_df['pred_confidence'] = y_prob_f1[mis_idx].max(axis=1)

    display(mis_df[['file_path', 'true_label', 'pred_label', 'pred_confidence']].head(10))

    n_show = min(6, len(mis_df))
    sample_mis = mis_df.sample(n=n_show, random_state=SEED).reset_index(drop=True)
    plt.figure(figsize=(14, 7))
    for i in range(n_show):
        img = tf.keras.utils.load_img(sample_mis.loc[i, 'file_path'], target_size=IMG_SIZE_F1)
        plt.subplot(2, 3, i + 1)
        plt.imshow(img)
        plt.title(
            f"Real: {sample_mis.loc[i, 'true_label']}\nPred: {sample_mis.loc[i, 'pred_label']}\nConf: {sample_mis.loc[i, 'pred_confidence']:.2f}"
        )
        plt.axis('off')
    plt.tight_layout()
    plt.show()

print('\nReporte de clasificación (macro):')
print(classification_report(y_test_f1, y_pred_f1, target_names=label_encoder_f1.classes_, zero_division=0))

cm = confusion_matrix(y_test_f1, y_pred_f1)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, cmap='Blues', cbar=False)
plt.title('Matriz de confusión (Fase 1)')
plt.xlabel('Predicción')
plt.ylabel('Real')
plt.tight_layout()
plt.show()
```

# Proyecto Final DL — Etapa 2 (Arquitectura Profunda Especializada)

## Propósito de la etapa
La Etapa 2 evalúa el aporte de una arquitectura convolucional profunda entrenada desde cero para clasificación multiclase de imágenes dermatológicas, con el objetivo de establecer una mejora técnica frente al baseline inicial y preparar la transición hacia modelos de mayor capacidad representacional.

## Alcance metodológico
En esta etapa se implementa una CNN profunda sin uso de pesos preentrenados, sin componentes Transformer y sin técnicas generativas, para aislar el efecto de una arquitectura visual especializada sobre el mismo problema y particiones de datos.

## Protocolo de evaluación
La evaluación se realiza mediante comparación directa con Etapa 1 usando métricas de clasificación (Accuracy, Precision macro, Recall macro y Macro F1). Como análisis complementario, se incorpora interpretación opcional con Grad-CAM para inspección cualitativa de regiones relevantes en la decisión del modelo.

```python

AUTOTUNE = tf.data.AUTOTUNE

IMG_SIZE_F2 = (224, 224)
BATCH_SIZE_F2 = 32
EPOCHS_F2 = 35
LR_F2 = 8e-4
WEIGHT_DECAY_F2 = 2e-4

if 'train_df' in globals() and 'val_df' in globals() and 'test_df' in globals() and 'label_encoder_f1' in globals():
    train_df_f2 = train_df.copy()
    val_df_f2 = val_df.copy()
    test_df_f2 = test_df.copy()
    label_encoder_f2 = label_encoder_f1
    num_classes_f2 = len(label_encoder_f2.classes_)
else:
    
    phase2_df = df[['file_path', 'label']].dropna().copy()
    class_counts_f2 = phase2_df['label'].value_counts()
    phase2_df = phase2_df[phase2_df['label'].isin(class_counts_f2[class_counts_f2 >= 5].index)].reset_index(drop=True)

    label_encoder_f2 = LabelEncoder()
    phase2_df['target'] = label_encoder_f2.fit_transform(phase2_df['label'])
    num_classes_f2 = len(label_encoder_f2.classes_)

    train_df_f2, temp_df_f2 = train_test_split(
        phase2_df,
        test_size=0.30,
        random_state=SEED,
        stratify=phase2_df['target']
    )
    val_df_f2, test_df_f2 = train_test_split(
        temp_df_f2,
        test_size=0.50,
        random_state=SEED,
        stratify=temp_df_f2['target']
    )

print('Etapa 2 - split train/val/test:', len(train_df_f2), len(val_df_f2), len(test_df_f2))
print('Etapa 2 - clases:', num_classes_f2)

classes_f2 = np.unique(train_df_f2['target'].values)
weights_f2 = compute_class_weight(
    class_weight='balanced',
    classes=classes_f2,
    y=train_df_f2['target'].values
)
weights_f2 = np.sqrt(weights_f2)
class_weight_f2 = {int(c): float(w) for c, w in zip(classes_f2, weights_f2)}

def decode_resize_f2(path, label):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE_F2, antialias=True)
    img = tf.cast(img, tf.float32)
    return img, label

def make_ds_f2(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(buffer_size=min(len(paths), 16000), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(decode_resize_f2, num_parallel_calls=AUTOTUNE)
    ds = ds.batch(BATCH_SIZE_F2)
    ds = ds.prefetch(AUTOTUNE)
    return ds

train_ds_f2 = make_ds_f2(train_df_f2['file_path'].values, train_df_f2['target'].values, training=True)
val_ds_f2 = make_ds_f2(val_df_f2['file_path'].values, val_df_f2['target'].values, training=False)
test_ds_f2 = make_ds_f2(test_df_f2['file_path'].values, test_df_f2['target'].values, training=False)

def conv_block(x, filters, drop=0.0):
    x = tf.keras.layers.Conv2D(filters, 3, padding='same', use_bias=False)(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Activation('relu')(x)
    x = tf.keras.layers.Conv2D(filters, 3, padding='same', use_bias=False)(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Activation('relu')(x)
    x = tf.keras.layers.MaxPooling2D(pool_size=2)(x)
    if drop > 0:
        x = tf.keras.layers.Dropout(drop)(x)
    return x

inputs_f2 = tf.keras.Input(shape=(IMG_SIZE_F2[0], IMG_SIZE_F2[1], 3))
x = tf.keras.layers.Rescaling(1.0 / 255.0)(inputs_f2)
x = tf.keras.layers.RandomFlip('horizontal')(x)
x = tf.keras.layers.RandomRotation(0.10)(x)
x = tf.keras.layers.RandomZoom(0.12)(x)
x = tf.keras.layers.RandomContrast(0.10)(x)

x = conv_block(x, 32, drop=0.10)
x = conv_block(x, 64, drop=0.15)
x = conv_block(x, 128, drop=0.20)
x = conv_block(x, 256, drop=0.25)

x = tf.keras.layers.Conv2D(320, 3, padding='same', use_bias=False, name='last_conv_f2')(x)
x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.Activation('relu')(x)

x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dense(256, activation='relu')(x)
x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.Dropout(0.45)(x)
outputs_f2 = tf.keras.layers.Dense(num_classes_f2, activation='softmax', dtype='float32')(x)

phase2_model = tf.keras.Model(inputs_f2, outputs_f2, name='phase2_cnn_scratch')

callbacks_f2 = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=8, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7),
]

optimizer_f2 = tf.keras.optimizers.AdamW(
    learning_rate=LR_F2,
    weight_decay=WEIGHT_DECAY_F2,
    clipnorm=1.0
)

phase2_model.compile(
    optimizer=optimizer_f2,
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print('\nEntrenamiento Etapa 2 (CNN desde cero)...')
history_f2 = phase2_model.fit(
    train_ds_f2,
    validation_data=val_ds_f2,
    epochs=EPOCHS_F2,
    class_weight=class_weight_f2,
    callbacks=callbacks_f2,
    verbose=1,
)

y_test_f2 = test_df_f2['target'].values
y_prob_f2 = phase2_model.predict(test_ds_f2, verbose=0)
y_pred_f2 = np.argmax(y_prob_f2, axis=1)

acc_f2 = accuracy_score(y_test_f2, y_pred_f2)
precision_macro_f2, recall_macro_f2, f1_macro_f2, _ = precision_recall_fscore_support(
    y_test_f2,
    y_pred_f2,
    average='macro',
    zero_division=0,
)

metrics_f2_df = pd.DataFrame({
    'metric': ['accuracy', 'precision_macro', 'recall_macro', 'macro_f1', 'target_goal'],
    'value': [acc_f2, precision_macro_f2, recall_macro_f2, f1_macro_f2, 0.60],
})

print(f'Etapa 2 - accuracy: {acc_f2:.4f}')
print(f'Etapa 2 - precision_macro: {precision_macro_f2:.4f}')
print(f'Etapa 2 - recall_macro: {recall_macro_f2:.4f}')
print(f'Etapa 2 - macro F1: {f1_macro_f2:.4f}')
print(f'Objetivo >= 0.60 cumplido: {acc_f2 >= 0.60 and f1_macro_f2 >= 0.60}')
display(metrics_f2_df)

if 'metrics_eval_df' in globals():
    phase1_metric_map = dict(zip(metrics_eval_df['metric'], metrics_eval_df['value']))
    acc_f1_cmp = float(phase1_metric_map.get('accuracy', np.nan))
    p_f1_cmp = float(phase1_metric_map.get('precision_macro', np.nan))
    r_f1_cmp = float(phase1_metric_map.get('recall_macro', np.nan))
    f1_f1_cmp = float(phase1_metric_map.get('macro_f1', np.nan))
else:
    acc_f1_cmp = np.nan
    p_f1_cmp = np.nan
    r_f1_cmp = np.nan
    f1_f1_cmp = np.nan
    print('Nota: no se encontró metrics_eval_df de Etapa 1 en memoria. Ejecuta Etapa 1 para comparación numérica completa.')

comparison_df = pd.DataFrame({
    'modelo': ['Etapa 1 - MLP base', 'Etapa 2 - CNN profunda'],
    'accuracy': [acc_f1_cmp, acc_f2],
    'precision_macro': [p_f1_cmp, precision_macro_f2],
    'recall_macro': [r_f1_cmp, recall_macro_f2],
    'macro_f1': [f1_f1_cmp, f1_macro_f2],
})
display(comparison_df)

plt.figure(figsize=(9, 5))
plot_df = comparison_df.melt(id_vars='modelo', var_name='metric', value_name='value')
sns.barplot(data=plot_df, x='metric', y='value', hue='modelo')
plt.axhline(0.60, color='green', linestyle='--', label='Meta 0.60')
plt.ylim(0, 1)
plt.title('Comparación cuantitativa: Etapa 1 vs Etapa 2')
plt.legend()
plt.tight_layout()
plt.show()


plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history_f2.history['loss'], label='train_loss')
plt.plot(history_f2.history['val_loss'], label='val_loss')
plt.title('Etapa 2 - Curvas de pérdida')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_f2.history['accuracy'], label='train_accuracy')
plt.plot(history_f2.history['val_accuracy'], label='val_accuracy')
plt.title('Etapa 2 - Curvas de accuracy')
plt.legend()
plt.tight_layout()
plt.show()
```

```python

if 'phase2_model' not in globals():
    raise RuntimeError('No se encontró phase2_model. Ejecuta primero la celda principal de Etapa 2.')


last_conv_name = 'last_conv_f2'

def make_gradcam_heatmap(model, img_tensor, conv_layer_name):
    conv_layer = model.get_layer(conv_layer_name)
    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[conv_layer.output, model.output],
    )

    with tf.GradientTape() as tape:
        conv_outputs, preds = grad_model([img_tensor], training=False)
        pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]

    grads = tape.gradient(class_channel, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_outputs = conv_outputs[0]
    heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)
    heatmap = tf.maximum(heatmap, 0) / (tf.reduce_max(heatmap) + 1e-8)
    return heatmap.numpy(), int(pred_index)

def overlay_heatmap_on_image(img_array, heatmap, alpha=0.35):
    heatmap = np.uint8(255 * heatmap)
    heatmap = tf.image.resize(heatmap[..., np.newaxis], (img_array.shape[0], img_array.shape[1])).numpy().squeeze()
    cmap = plt.get_cmap('jet')
    color_heatmap = cmap(heatmap / 255.0)[..., :3]
    superimposed = np.clip((img_array / 255.0) * (1 - alpha) + color_heatmap * alpha, 0, 1)
    return superimposed

sample_paths = test_df_f2['file_path'].sample(n=min(4, len(test_df_f2)), random_state=SEED).tolist()

plt.figure(figsize=(14, 8))
for i, pth in enumerate(sample_paths, start=1):
    pil_img = tf.keras.utils.load_img(pth, target_size=IMG_SIZE_F2)
    arr = tf.keras.utils.img_to_array(pil_img).astype(np.float32)
    input_tensor = np.expand_dims(arr, axis=0)

    heatmap, pred_idx = make_gradcam_heatmap(phase2_model, input_tensor, last_conv_name)
    overlay = overlay_heatmap_on_image(arr, heatmap, alpha=0.35)

    pred_label = label_encoder_f2.inverse_transform([pred_idx])[0]

    plt.subplot(2, 4, i)
    plt.imshow(arr.astype(np.uint8))
    plt.title(f'Original\nPred: {pred_label}')
    plt.axis('off')

    plt.subplot(2, 4, i + 4)
    plt.imshow(overlay)
    plt.title('Grad-CAM')
    plt.axis('off')

plt.suptitle('Etapa 2 [Opcional] - Mapas de activación Grad-CAM', y=1.02)
plt.tight_layout()
plt.show()
```

## Protocolo de evaluación

Adicionalmente, la etapa integra análisis de costo computacional frente a ganancia de desempeño para sustentar la selección del modelo candidato a etapa posterior.

```python

AUTOTUNE = tf.data.AUTOTUNE

vit_model_label = 'ViT-B16 (TF-Hub)'
try:
    import tensorflow_hub as hub
except Exception:
    import sys
    import subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'tensorflow-hub'])
    import tensorflow_hub as hub

IMG_SIZE_F3 = (224, 224)
BATCH_SIZE_F3 = 32
EPOCHS_EFF_F3 = 12
EPOCHS_VIT_F3 = 8
LR_EFF_F3 = 3e-4
LR_VIT_F3 = 2e-4
WEIGHT_DECAY_F3 = 1e-4

if 'train_df_f2' in globals() and 'val_df_f2' in globals() and 'test_df_f2' in globals() and 'label_encoder_f2' in globals():
    train_df_f3 = train_df_f2.copy()
    val_df_f3 = val_df_f2.copy()
    test_df_f3 = test_df_f2.copy()
    label_encoder_f3 = label_encoder_f2
    num_classes_f3 = len(label_encoder_f3.classes_)
elif 'train_df' in globals() and 'val_df' in globals() and 'test_df' in globals() and 'label_encoder_f1' in globals():
    train_df_f3 = train_df.copy()
    val_df_f3 = val_df.copy()
    test_df_f3 = test_df.copy()
    label_encoder_f3 = label_encoder_f1
    num_classes_f3 = len(label_encoder_f3.classes_)
else:
    phase3_df = df[['file_path', 'label']].dropna().copy()
    class_counts_f3 = phase3_df['label'].value_counts()
    phase3_df = phase3_df[phase3_df['label'].isin(class_counts_f3[class_counts_f3 >= 5].index)].reset_index(drop=True)

    label_encoder_f3 = LabelEncoder()
    phase3_df['target'] = label_encoder_f3.fit_transform(phase3_df['label'])
    num_classes_f3 = len(label_encoder_f3.classes_)

    train_df_f3, temp_df_f3 = train_test_split(
        phase3_df,
        test_size=0.30,
        random_state=SEED,
        stratify=phase3_df['target']
    )
    val_df_f3, test_df_f3 = train_test_split(
        temp_df_f3,
        test_size=0.50,
        random_state=SEED,
        stratify=temp_df_f3['target']
    )

print('Etapa 3 - split train/val/test:', len(train_df_f3), len(val_df_f3), len(test_df_f3))
print('Etapa 3 - clases:', num_classes_f3)
print('Modelo Transformer usado:', vit_model_label)

classes_f3 = np.unique(train_df_f3['target'].values)
weights_f3 = compute_class_weight(
    class_weight='balanced',
    classes=classes_f3,
    y=train_df_f3['target'].values
)
weights_f3 = np.sqrt(weights_f3)
class_weight_f3 = {int(c): float(w) for c, w in zip(classes_f3, weights_f3)}

def decode_resize_f3(path, label):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img = tf.image.resize(img, IMG_SIZE_F3, antialias=True)
    img = tf.cast(img, tf.float32)
    return img, label

def make_ds_f3(paths, labels, training=False):
    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(buffer_size=min(len(paths), 16000), seed=SEED, reshuffle_each_iteration=True)
    ds = ds.map(decode_resize_f3, num_parallel_calls=AUTOTUNE)
    ds = ds.batch(BATCH_SIZE_F3)
    ds = ds.prefetch(AUTOTUNE)
    return ds

train_ds_f3 = make_ds_f3(train_df_f3['file_path'].values, train_df_f3['target'].values, training=True)
val_ds_f3 = make_ds_f3(val_df_f3['file_path'].values, val_df_f3['target'].values, training=False)
test_ds_f3 = make_ds_f3(test_df_f3['file_path'].values, test_df_f3['target'].values, training=False)

base_model_eff = tf.keras.applications.EfficientNetB0(
    include_top=False,
    weights='imagenet',
    input_shape=(IMG_SIZE_F3[0], IMG_SIZE_F3[1], 3),
)
base_model_eff.trainable = False

inputs_eff = tf.keras.Input(shape=(IMG_SIZE_F3[0], IMG_SIZE_F3[1], 3))
x = tf.keras.layers.RandomFlip('horizontal')(inputs_eff)
x = tf.keras.layers.RandomRotation(0.10)(x)
x = tf.keras.layers.RandomZoom(0.10)(x)
x = tf.keras.layers.RandomContrast(0.10)(x)
x = base_model_eff(x, training=False)
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.Dense(256, activation='relu')(x)
x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.Dropout(0.40)(x)
outputs_eff = tf.keras.layers.Dense(num_classes_f3, activation='softmax', dtype='float32')(x)

model_eff = tf.keras.Model(inputs_eff, outputs_eff, name='phase3_efficientnetb0')
model_eff.compile(
    optimizer=tf.keras.optimizers.AdamW(learning_rate=LR_EFF_F3, weight_decay=WEIGHT_DECAY_F3, clipnorm=1.0),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

callbacks_eff = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=4, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, min_lr=1e-7),
]

print('\nEntrenamiento EfficientNetB0...')
history_eff = model_eff.fit(
    train_ds_f3,
    validation_data=val_ds_f3,
    epochs=EPOCHS_EFF_F3,
    class_weight=class_weight_f3,
    callbacks=callbacks_eff,
    verbose=1,
)

vit_encoder = hub.KerasLayer('https://tfhub.dev/sayakpaul/vit_b16_fe/1', trainable=False)

def _select_vit_tensor(vit_output):
    if isinstance(vit_output, dict):
        if 'pooled_output' in vit_output:
            return vit_output['pooled_output']
        if 'sequence_output' in vit_output:
            return vit_output['sequence_output']
        return next(iter(vit_output.values()))
    return vit_output

def extract_vit_features(ds):
    feat_batches, label_batches = [], []
    for xb, yb in ds:
        xb = tf.cast(xb, tf.float32) / 255.0
        vit_out = vit_encoder(xb)
        vit_out = _select_vit_tensor(vit_out)
        if len(vit_out.shape) == 3:
            vit_out = tf.reduce_mean(vit_out, axis=1)
        feat_batches.append(vit_out.numpy().astype(np.float32))
        label_batches.append(yb.numpy().astype(np.int32))
    return np.concatenate(feat_batches, axis=0), np.concatenate(label_batches, axis=0)

print('\nExtrayendo embeddings ViT (train/val/test)...')
x_train_vit, y_train_vit = extract_vit_features(train_ds_f3)
x_val_vit, y_val_vit = extract_vit_features(val_ds_f3)
x_test_vit, y_test_vit = extract_vit_features(test_ds_f3)

model_vit = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(x_train_vit.shape[1],)),
    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dropout(0.35),
    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.Dropout(0.25),
    tf.keras.layers.Dense(num_classes_f3, activation='softmax', dtype='float32'),
] , name='phase3_vit_head')

model_vit.compile(
    optimizer=tf.keras.optimizers.AdamW(learning_rate=LR_VIT_F3, weight_decay=WEIGHT_DECAY_F3, clipnorm=1.0),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

callbacks_vit = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=1, min_lr=1e-7),
]

print(f'\nEntrenamiento cabeza de clasificación sobre embeddings {vit_model_label}...')
history_vit = model_vit.fit(
    x_train_vit,
    y_train_vit,
    validation_data=(x_val_vit, y_val_vit),
    epochs=EPOCHS_VIT_F3,
    class_weight=class_weight_f3,
    callbacks=callbacks_vit,
    batch_size=256,
    verbose=1,
)

def evaluate_classifier(model, x_eval, y_true, from_logits=False):
    y_out = model.predict(x_eval, verbose=0)
    if from_logits:
        y_out = tf.nn.softmax(y_out, axis=1).numpy()
    y_pred = np.argmax(y_out, axis=1)
    acc = accuracy_score(y_true, y_pred)
    p_macro, r_macro, f1_macro, _ = precision_recall_fscore_support(
        y_true, y_pred, average='macro', zero_division=0
    )
    return acc, p_macro, r_macro, f1_macro

y_test_f3 = test_df_f3['target'].values
acc_eff, p_eff, r_eff, f1_eff = evaluate_classifier(model_eff, test_ds_f3, y_test_f3, from_logits=False)
acc_vit, p_vit, r_vit, f1_vit = evaluate_classifier(model_vit, x_test_vit, y_test_vit, from_logits=False)

metrics_pretrained_df = pd.DataFrame({
    'modelo_preentrenado': ['EfficientNetB0', vit_model_label],
    'accuracy': [acc_eff, acc_vit],
    'precision_macro': [p_eff, p_vit],
    'recall_macro': [r_eff, r_vit],
    'macro_f1': [f1_eff, f1_vit],
})
display(metrics_pretrained_df)


best_row = metrics_pretrained_df.sort_values('macro_f1', ascending=False).iloc[0]
acc_f3 = float(best_row['accuracy'])
precision_macro_f3 = float(best_row['precision_macro'])
recall_macro_f3 = float(best_row['recall_macro'])
f1_macro_f3 = float(best_row['macro_f1'])
best_model_name_f3 = best_row['modelo_preentrenado']

metrics_f3_df = pd.DataFrame({
    'metric': ['accuracy', 'precision_macro', 'recall_macro', 'macro_f1', 'target_goal'],
    'value': [acc_f3, precision_macro_f3, recall_macro_f3, f1_macro_f3, 0.60],
})

print(f"Etapa 3 - mejor modelo preentrenado: {best_model_name_f3}")
print(f'Etapa 3 - accuracy: {acc_f3:.4f}')
print(f'Etapa 3 - precision_macro: {precision_macro_f3:.4f}')
print(f'Etapa 3 - recall_macro: {recall_macro_f3:.4f}')
print(f'Etapa 3 - macro F1: {f1_macro_f3:.4f}')
print(f'Objetivo >= 0.60 cumplido: {acc_f3 >= 0.60 and f1_macro_f3 >= 0.60}')
display(metrics_f3_df)

def _metric_lookup(metric_df, metric_name):
    if 'metric' not in metric_df.columns:
        return np.nan
    match = metric_df.loc[metric_df['metric'] == metric_name, 'value']
    return float(match.iloc[0]) if len(match) else np.nan

acc_f1_cmp = _metric_lookup(metrics_eval_df, 'accuracy') if 'metrics_eval_df' in globals() else np.nan
p_f1_cmp = _metric_lookup(metrics_eval_df, 'precision_macro') if 'metrics_eval_df' in globals() else np.nan
r_f1_cmp = _metric_lookup(metrics_eval_df, 'recall_macro') if 'metrics_eval_df' in globals() else np.nan
f1_f1_cmp = _metric_lookup(metrics_eval_df, 'macro_f1') if 'metrics_eval_df' in globals() else np.nan

acc_f2_cmp = _metric_lookup(metrics_f2_df, 'accuracy') if 'metrics_f2_df' in globals() else np.nan
p_f2_cmp = _metric_lookup(metrics_f2_df, 'precision_macro') if 'metrics_f2_df' in globals() else np.nan
r_f2_cmp = _metric_lookup(metrics_f2_df, 'recall_macro') if 'metrics_f2_df' in globals() else np.nan
f1_f2_cmp = _metric_lookup(metrics_f2_df, 'macro_f1') if 'metrics_f2_df' in globals() else np.nan

comparison_stage_df = pd.DataFrame({
    'modelo': ['Etapa 1 - MLP base', 'Etapa 2 - CNN profunda', f'Etapa 3 - {best_model_name_f3}'],
    'accuracy': [acc_f1_cmp, acc_f2_cmp, acc_f3],
    'precision_macro': [p_f1_cmp, p_f2_cmp, precision_macro_f3],
    'recall_macro': [r_f1_cmp, r_f2_cmp, recall_macro_f3],
    'macro_f1': [f1_f1_cmp, f1_f2_cmp, f1_macro_f3],
})
display(comparison_stage_df)

plt.figure(figsize=(11, 5))
plot_df_f3 = comparison_stage_df.melt(id_vars='modelo', var_name='metric', value_name='value')
sns.barplot(data=plot_df_f3, x='metric', y='value', hue='modelo')
plt.axhline(0.60, color='green', linestyle='--', label='Meta 0.60')
plt.ylim(0, 1)
plt.title('Comparación cuantitativa: Etapas 1, 2 y 3')
plt.legend()
plt.tight_layout()
plt.show()

def count_params_of_model(model):
    trainable = int(np.sum([tf.keras.backend.count_params(w) for w in model.trainable_weights]))
    non_trainable = int(np.sum([tf.keras.backend.count_params(w) for w in model.non_trainable_weights]))
    return trainable, non_trainable, trainable + non_trainable

eff_trainable, eff_non_trainable, eff_total = count_params_of_model(model_eff)
vit_head_trainable, vit_head_non_trainable, _ = count_params_of_model(model_vit)
vit_encoder_params = int(np.sum([np.prod(w.shape) for w in vit_encoder.weights]))
vit_trainable = vit_head_trainable
vit_non_trainable = vit_head_non_trainable + vit_encoder_params
vit_total = vit_trainable + vit_non_trainable
f1_ref_f2 = _metric_lookup(metrics_f2_df, 'macro_f1') if 'metrics_f2_df' in globals() else np.nan

cost_gain_df = pd.DataFrame({
    'modelo_preentrenado': ['EfficientNetB0', vit_model_label],
    'trainable_params': [eff_trainable, vit_trainable],
    'non_trainable_params': [eff_non_trainable, vit_non_trainable],
    'total_params': [eff_total, vit_total],
    'macro_f1': [f1_eff, f1_vit],
    'delta_macro_f1_vs_etapa2': [f1_eff - f1_ref_f2, f1_vit - f1_ref_f2],
})
display(cost_gain_df)


plt.figure(figsize=(14, 4))
plt.subplot(1, 2, 1)
plt.plot(history_eff.history['val_accuracy'], label='val_acc EfficientNetB0')
plt.plot(history_vit.history['val_accuracy'], label=f'val_acc {vit_model_label}')
plt.title('Etapa 3 - Validación accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history_eff.history['val_loss'], label='val_loss EfficientNetB0')
plt.plot(history_vit.history['val_loss'], label=f'val_loss {vit_model_label}')
plt.title('Etapa 3 - Validación loss')
plt.legend()
plt.tight_layout()
plt.show()
```
